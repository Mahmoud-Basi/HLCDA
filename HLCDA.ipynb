```python
# ============================================================
# APT Detection (Sequences, T=8)
# Models: CNN+BiLSTM, CNN+xLSTM, CNN+Co-Attention, HLCDA
# Outputs: Confusion Matrix, Training Curves, Reliability, t-SNE
# (UPDATED) + AUROC/AUPRC + latency + optimizer choice + LR logging
# (NEW)     + EVASION / ROBUSTNESS EVALUATION (feature-space)
# ============================================================

# --- Libraries ---
import os, json, time, random
from collections import Counter

import numpy as np
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader

# --- Plotting libs and pro styles ---
import matplotlib
import seaborn as sns
import matplotlib.pyplot as plt

from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score, confusion_matrix,
    roc_auc_score, average_precision_score
)
from sklearn.manifold import TSNE


# -------------------- GLOBAL STYLE --------------------
def set_global_fonts(base=14, title=18, label=14, ticks=12, legend=12):
    matplotlib.rcParams.update({
        "font.size": base,
        "axes.titlesize": title,
        "axes.labelsize": label,
        "xtick.labelsize": ticks,
        "ytick.labelsize": ticks,
        "legend.fontsize": legend,
        "figure.titlesize": title + 2,
    })

set_global_fonts(base=14, title=18, label=14, ticks=12, legend=12)

matplotlib.rcParams.update({
    "font.size": 14,
    "axes.titlesize": 16,
    "axes.labelsize": 14,
    "xtick.labelsize": 12,
    "ytick.labelsize": 12,
    "legend.fontsize": 12,
    "figure.titlesize": 18
})
sns.set_style("whitegrid")


# -------------------- CONFIG --------------------
NPZ_PATH     = r"C:\Users\LENOVO\Jupyter_Files\dataset\prepped\sessions_T8.npz"
DEVICE       = "cuda" if torch.cuda.is_available() else "cpu"
SEQ_BATCH    = 256
SEQ_EPOCHS   = 100
LR           = 1e-3
WEIGHT_DECAY = 1e-4
SEED         = 42

# Optimizer selection
OPT_NAME = "AdamW"  # "AdamW" | "Adam" | "SGD"

# -------------------- (A) LR SCHEDULER: COSINE --------------------
SCHED_NAME = "cosine"   # "cosine" | "none"
ETA_MIN    = 1e-5       # final LR for cosine annealing

# -------------------- (NEW) EVASION / ROBUSTNESS CONFIG --------------------
RUN_EVASION_EVAL = True

# How strong the evasion is
EVASION_CFG = {
    "timing_jitter": {"alpha": 0.30},          # noise strength on timing features
    "rate_shaping":  {"alpha": 0.25},          # scale down rate features
    "size_frag_pad": {"alpha": 0.20},          # scale packet/byte size features
    "benign_drift":  {"beta":  0.25},          # pull features toward benign mean
    "mixed":         {"alpha": 0.12, "beta": 0.15},  # mild mix
}

# Reproducibility
random.seed(SEED); np.random.seed(SEED); torch.manual_seed(SEED)
if DEVICE == "cuda":
    torch.cuda.manual_seed_all(SEED)
print("[INFO] DEVICE =", DEVICE)

# --- Scenario Color Palettes ---
SCENARIO_COLORS = {
    "CNN+BiLSTM": ("Purples", ["#756bb1", "#bcbddc"]),
    "CNN+xLSTM":  ("Blues",   ["#1f77b4", "#aec7e8"]),
    "CNN+CoAttn": ("Greens",  ["#2ca02c", "#98df8a"]),
    "HLCDA":      ("Oranges", ["#ff7f0e", "#ffbb78"])
}


# -------------------- DATA ----------------------
def load_npz(npz_path):
    d = np.load(npz_path, allow_pickle=True)
    Xtr, ytr, Ltr = d["X_train"], d["y_train"], d["len_train"]
    Xva, yva, Lva = d["X_val"],   d["y_val"],   d["len_val"]
    Xte, yte, Lte = d["X_test"],  d["y_test"],  d["len_test"]
    feats = d["features"].tolist()
    T     = int(d["T"][0])
    return (Xtr,ytr,Ltr), (Xva,yva,Lva), (Xte,yte,Lte), feats, T

(train_t, val_t, test_t, feature_names, T) = load_npz(NPZ_PATH)
Xtr,ytr,Ltr = train_t; Xva,yva,Lva = val_t; Xte,yte,Lte = test_t
F = Xtr.shape[-1]
print(f"[SEQ] Train {Xtr.shape}, Val {Xva.shape}, Test {Xte.shape}, F={F}, T={T}")

class SeqDS(Dataset):
    def __init__(self, X, y, L):
        self.X = torch.from_numpy(X).float()
        self.y = torch.from_numpy(y).long()
        self.L = torch.from_numpy(L).long()
    def __len__(self): return self.X.shape[0]
    def __getitem__(self, i): return self.X[i], self.y[i], self.L[i]

tr_ds, va_ds, te_ds = SeqDS(Xtr,ytr,Ltr), SeqDS(Xva,yva,Lva), SeqDS(Xte,yte,Lte)

# GPU-friendly DataLoader knobs
PIN = (DEVICE == "cuda")
tr_loader = DataLoader(tr_ds, batch_size=SEQ_BATCH, shuffle=True,  num_workers=0, pin_memory=PIN)
va_loader = DataLoader(va_ds, batch_size=SEQ_BATCH, shuffle=False, num_workers=0, pin_memory=PIN)
te_loader = DataLoader(te_ds, batch_size=SEQ_BATCH, shuffle=False, num_workers=0, pin_memory=PIN)

# Class weights (binary)
ctr = Counter(ytr.tolist())
w0 = len(ytr) / (2.0 * ctr.get(0,1))
w1 = len(ytr) / (2.0 * ctr.get(1,1))
class_weights = torch.tensor([w0, w1], dtype=torch.float32, device=DEVICE)
print("[SEQ] class_weights =", class_weights.tolist())


# -------------------- OPTIMIZER FACTORY --------------------
def build_optimizer(name, params, lr, weight_decay):
    name = name.lower()
    if name == "adamw":
        return torch.optim.AdamW(params, lr=lr, weight_decay=weight_decay)
    if name == "adam":
        return torch.optim.Adam(params, lr=lr, weight_decay=weight_decay)
    if name == "sgd":
        return torch.optim.SGD(params, lr=lr, momentum=0.9, nesterov=True, weight_decay=weight_decay)
    raise ValueError(f"Unknown optimizer: {name}")


# -------------------- BUILDING BLOCKS --------------------
class CNNFront(nn.Module):
    """Parallel convs (k=3,5,7) -> concat to EXACT out_dim."""
    def __init__(self, in_dim, out_dim=128, drop=0.2):
        super().__init__()
        ks = [3,5,7]
        base = out_dim // len(ks)
        rem  = out_dim - base * len(ks)
        branch_outs = [base + (1 if i < rem else 0) for i in range(len(ks))]
        self.branches = nn.ModuleList([
            nn.Sequential(
                nn.Conv1d(in_dim, bo, k, padding=k//2),
                nn.BatchNorm1d(bo),
                nn.ReLU(),
                nn.Dropout(drop)
            ) for bo, k in zip(branch_outs, ks)
        ])
        self.out_dim = out_dim

    def forward(self, x):                # x: [B,T,F]
        x = x.transpose(1,2)             # [B,F,T]
        feats = torch.cat([b(x) for b in self.branches], dim=1)   # [B,out_dim,T]
        return feats.transpose(1,2)      # [B,T,out_dim]

class BiLSTMHead(nn.Module):
    def __init__(self, in_dim, hid=128, layers=1, drop=0.3, bidir=True):
        super().__init__()
        self.rnn = nn.LSTM(in_dim, hid, num_layers=layers, batch_first=True,
                           dropout=drop if layers>1 else 0.0, bidirectional=bidir)
        self.out_dim = hid*(2 if bidir else 1)
    def forward(self, x, lengths=None):
        self.rnn.flatten_parameters()
        y,_ = self.rnn(x)               # [B,T,out_dim]
        return y

class xLSTMHead(nn.Module):
    def __init__(self, in_dim, hid=192, proj=128, layers=1, drop=0.3, bidir=True):
        super().__init__()
        self.rnn = nn.LSTM(in_dim, hid, num_layers=layers, batch_first=True,
                           dropout=drop if layers>1 else 0.0, bidirectional=bidir)
        self.ln  = nn.LayerNorm(hid*(2 if bidir else 1))
        self.proj= nn.Linear(hid*(2 if bidir else 1), proj)
        self.out_dim = proj
    def forward(self, x, lengths=None):
        self.rnn.flatten_parameters()
        y,_ = self.rnn(x)               # [B,T,2*hid]
        y = self.proj(self.ln(y))       # [B,T,proj]
        return y

class CoAttentionBlock(nn.Module):
    def __init__(self, dim_s, dim_u, heads=4, ffn=256, drop=0.1):
        super().__init__()
        self.att_s_from_u = nn.MultiheadAttention(embed_dim=dim_s, kdim=dim_u, vdim=dim_u,
                                                  num_heads=heads, batch_first=True, dropout=drop)
        self.att_u_from_s = nn.MultiheadAttention(embed_dim=dim_u, kdim=dim_s, vdim=dim_s,
                                                  num_heads=heads, batch_first=True, dropout=drop)
        self.ln_s = nn.LayerNorm(dim_s); self.ln_u = nn.LayerNorm(dim_u)
        self.ffn_s = nn.Sequential(nn.Linear(dim_s, ffn), nn.ReLU(), nn.Dropout(drop), nn.Linear(ffn, dim_s))
        self.ffn_u = nn.Sequential(nn.Linear(dim_u, ffn), nn.ReLU(), nn.Dropout(drop), nn.Linear(ffn, dim_u))
    def forward(self, S, U, key_padding_mask=None):
        # Ensure bool mask (True = masked/pad)
        if key_padding_mask is not None:
            key_padding_mask = key_padding_mask.bool()

        Sn = self.ln_s(S); Un = self.ln_u(U)
        S2,_ = self.att_s_from_u(Sn, Un, Un, key_padding_mask=key_padding_mask)
        U2,_ = self.att_u_from_s(Un, Sn, Sn, key_padding_mask=key_padding_mask)
        S = S + S2; U = U + U2
        S = S + self.ffn_s(self.ln_s(S))
        U = U + self.ffn_u(self.ln_u(U))
        return S, U

class AttnPool(nn.Module):
    def __init__(self, dim):
        super().__init__()
        self.w = nn.Linear(dim, 1)
    def forward(self, x, lengths=None, mask=None):
        scores = self.w(x).squeeze(-1)           # [B,T]
        if mask is not None:
            scores = scores.masked_fill(mask, float("-inf"))
        attn = torch.softmax(scores, dim=-1)     # [B,T]
        pooled = torch.sum(x * attn.unsqueeze(-1), dim=1)  # [B,D]
        return pooled, attn

def make_pad_mask(lengths, T):
    lengths = lengths.clamp(min=1, max=T)
    idx = torch.arange(T, device=lengths.device).unsqueeze(0)
    return idx >= lengths.unsqueeze(1)          # True = PAD


# -------------------- MODELS --------------------
class CNN_BiLSTM(nn.Module):
    def __init__(self, F, cnn_dim=128, rnn_hid=128, out_dim=2):
        super().__init__()
        self.cnn = CNNFront(F, out_dim=cnn_dim)
        self.rnn = BiLSTMHead(cnn_dim, hid=rnn_hid, layers=1, bidir=True)
        self.pool= AttnPool(self.rnn.out_dim)
        self.head= nn.Sequential(nn.Linear(self.rnn.out_dim, 128), nn.ReLU(), nn.Dropout(0.3),
                                 nn.Linear(128, out_dim))
    def forward(self, x, lengths):
        S = self.cnn(x)
        U = self.rnn(S, lengths)
        mask = make_pad_mask(lengths, x.size(1))
        P,_ = self.pool(U, lengths, mask)
        return self.head(P)

class CNN_xLSTM(nn.Module):
    def __init__(self, F, cnn_dim=128, rnn_hid=192, proj=128, out_dim=2):
        super().__init__()
        self.cnn = CNNFront(F, out_dim=cnn_dim)
        self.rnn = xLSTMHead(cnn_dim, hid=rnn_hid, proj=proj, layers=1, bidir=True)
        self.pool= AttnPool(self.rnn.out_dim)
        self.head= nn.Sequential(nn.Linear(self.rnn.out_dim, 128), nn.ReLU(), nn.Dropout(0.3),
                                 nn.Linear(128, out_dim))
    def forward(self, x, lengths):
        S = self.cnn(x)
        U = self.rnn(S, lengths)
        mask = make_pad_mask(lengths, x.size(1))
        P,_ = self.pool(U, lengths, mask)
        return self.head(P)

class CNN_CoAttn(nn.Module):
    def __init__(self, F, cnn_dim=128, rnn_hid=192, proj=128, heads=4, out_dim=2):
        super().__init__()
        self.cnn = CNNFront(F, out_dim=cnn_dim)
        self.rnn = xLSTMHead(cnn_dim, hid=rnn_hid, proj=proj, layers=1, bidir=True)
        self.coa = CoAttentionBlock(dim_s=cnn_dim, dim_u=self.rnn.out_dim, heads=heads, ffn=256, drop=0.1)
        self.pool_s = AttnPool(cnn_dim); self.pool_u = AttnPool(self.rnn.out_dim)
        self.head = nn.Sequential(nn.Linear(cnn_dim + self.rnn.out_dim, 256), nn.ReLU(), nn.Dropout(0.3),
                                  nn.Linear(256, out_dim))
    def forward(self, x, lengths):
        S = self.cnn(x)
        U = self.rnn(S, lengths)
        mask = make_pad_mask(lengths, x.size(1))
        S,U = self.coa(S, U, key_padding_mask=mask)
        Sp,_ = self.pool_s(S, lengths, mask)
        Up,_ = self.pool_u(U, lengths, mask)
        return self.head(torch.cat([Sp,Up], dim=-1))


# -------------------- RoPE + Transformer + DMFF for HLCDA --------------------
class RoPE1D(nn.Module):
    def __init__(self, dim):
        super().__init__()
        self.dim = dim
        inv_freq = 1.0 / (10000 ** (torch.arange(0, dim, 2).float() / dim))
        self.register_buffer("inv_freq", inv_freq, persistent=False)

    def _rope(self, x):
        B, T, H = x.shape
        t = torch.arange(T, device=x.device, dtype=self.inv_freq.dtype)
        freqs = torch.einsum("t,d->td", t, self.inv_freq)
        emb = torch.cat((freqs, freqs), dim=-1)
        cos = emb.cos()[None, ...]
        sin = emb.sin()[None, ...]
        x1, x2 = x[..., ::2], x[..., 1::2]
        x_rope = torch.stack([x1 * cos[..., ::2] - x2 * sin[..., ::2],
                              x1 * sin[..., ::2] + x2 * cos[..., ::2]], dim=-1)
        return x_rope.flatten(-2)

    def apply_rope(self, q, k):
        if q.size(-1) % 2 != 0:
            q = torch.nn.functional.pad(q, (0, 1))
            k = torch.nn.functional.pad(k, (0, 1))
        return self._rope(q), self._rope(k)

class SimpleTransformerBlock(nn.Module):
    def __init__(self, d_model=128, nhead=4, dim_ff=256, dropout=0.1, causal=True):
        super().__init__()
        self.causal = causal
        self.attn = nn.MultiheadAttention(d_model, nhead, batch_first=True, dropout=dropout)
        self.ffn = nn.Sequential(
            nn.Linear(d_model, dim_ff), nn.GELU(), nn.Dropout(dropout),
            nn.Linear(dim_ff, d_model)
        )
        self.ln1 = nn.LayerNorm(d_model)
        self.ln2 = nn.LayerNorm(d_model)
        self.rope = RoPE1D(d_model)

    def forward(self, x, key_padding_mask=None):
        # Ensure bool padding mask
        if key_padding_mask is not None:
            key_padding_mask = key_padding_mask.bool()

        x1 = self.ln1(x)
        q = k = x1
        q_rope, k_rope = self.rope.apply_rope(q, k)

        # ✅ FIX: bool causal mask (True = block future positions)
        attn_mask = None
        if self.causal:
            T_ = x.size(1)
            attn_mask = torch.triu(
                torch.ones((T_, T_), device=x.device, dtype=torch.bool),
                diagonal=1
            )

        y, _ = self.attn(
            q_rope, k_rope, x1,
            key_padding_mask=key_padding_mask,
            attn_mask=attn_mask
        )

        x = x + y
        x = x + self.ffn(self.ln2(x))
        return x

class DMFF(nn.Module):
    def __init__(self, in_dims, out_dim=256, heads=4):
        super().__init__()
        self.K = len(in_dims)
        self.H = heads
        self.Wk = nn.ModuleList([nn.ModuleList([nn.Linear(in_dims[k], out_dim) for k in range(self.K)])
                                 for _ in range(self.H)])
        self.vo = nn.ModuleList([nn.Linear(out_dim, out_dim, bias=False) for _ in range(self.H)])
        self.vq = nn.ParameterList([nn.Parameter(torch.randn(out_dim)) for _ in range(self.H)])

    def forward(self, feats):
        out = 0.0
        for h in range(self.H):
            proj_k = [self.Wk[h][k](feats[k]) for k in range(self.K)]
            scores = torch.stack([torch.einsum("bd,d->b", pk, self.vq[h]) for pk in proj_k], dim=1)
            gate = torch.softmax(scores, dim=1)
            mix = torch.stack(proj_k, dim=1)
            mixed = torch.einsum("bk,bkd->bd", gate, mix)
            out = out + self.vo[h](mixed)
        return out

class HLCDA(nn.Module):
    def __init__(self, F, cnn_dim=128, rnn_hid=192, proj=128,
                 heads_coa=4, coa_layers=2, d_model=128, tr_heads=4, tr_ff=256,
                 dmff_heads=4, out_dim=2):
        super().__init__()
        self.cnn = CNNFront(F, out_dim=cnn_dim)
        self.xlstm = xLSTMHead(cnn_dim, hid=rnn_hid, proj=proj, layers=1, bidir=True)

        self.coa_layers = nn.ModuleList(
            [CoAttentionBlock(dim_s=cnn_dim, dim_u=self.xlstm.out_dim,
                              heads=heads_coa, ffn=256, drop=0.1)
             for _ in range(coa_layers)]
        )

        self.fuse_proj = nn.Linear(cnn_dim + self.xlstm.out_dim, d_model)
        self.tr_block  = SimpleTransformerBlock(d_model=d_model, nhead=tr_heads, dim_ff=tr_ff, dropout=0.1, causal=True)
        self.ln_tr     = nn.LayerNorm(d_model)

        self.pool_s = AttnPool(cnn_dim)
        self.pool_u = AttnPool(self.xlstm.out_dim)

        self.dmff = DMFF(in_dims=[self.xlstm.out_dim, cnn_dim, d_model], out_dim=256, heads=dmff_heads)

        self.head = nn.Sequential(
            nn.Linear(256, 256), nn.GELU(), nn.Dropout(0.3),
            nn.Linear(256, out_dim)
        )

    def forward(self, x, lengths):
        S = self.cnn(x)
        U = self.xlstm(S, lengths)

        mask = make_pad_mask(lengths, x.size(1))
        for blk in self.coa_layers:
            S, U = blk(S, U, key_padding_mask=mask)

        Sp, _ = self.pool_s(S, lengths, mask)
        Up, _ = self.pool_u(U, lengths, mask)

        Z = torch.cat([S, U], dim=-1)
        Z = self.fuse_proj(Z)
        Z = self.tr_block(Z, key_padding_mask=mask)
        Z = self.ln_tr(Z)
        Tr = Z.mean(dim=1)

        Ff = self.dmff([Up, Sp, Tr])
        return self.head(Ff)


# -------------------- STYLING UTILS --------------------
def style_axes(ax,
               title_bold=True, label_bold=False, tick_bold=False, legend_bold=False,
               title_size=None, label_size=None, tick_size=None, legend_size=None):
    if ax.get_title():
        ax.set_title(ax.get_title(),
                     fontweight="bold" if title_bold else "normal",
                     fontsize=title_size or matplotlib.rcParams["axes.titlesize"])
    if ax.get_xlabel():
        ax.set_xlabel(ax.get_xlabel(),
                      fontweight="bold" if label_bold else "normal",
                      fontsize=label_size or matplotlib.rcParams["axes.labelsize"])
    if ax.get_ylabel():
        ax.set_ylabel(ax.get_ylabel(),
                      fontweight="bold" if label_bold else "normal",
                      fontsize=label_size or matplotlib.rcParams["axes.labelsize"])

    for ticklab in ax.get_xticklabels() + ax.get_yticklabels():
        ticklab.set_fontsize(tick_size or matplotlib.rcParams["xtick.labelsize"])
        ticklab.set_fontweight("bold" if tick_bold else "normal")

    leg = ax.get_legend()
    if leg is not None:
        if legend_size is not None:
            plt.setp(leg.get_texts(), fontsize=legend_size)
            if leg.get_title() is not None:
                leg.get_title().set_fontsize(legend_size)
        if legend_bold:
            if leg.get_title() is not None:
                leg.get_title().set_fontweight("bold")
            for txt in leg.get_texts():
                txt.set_fontweight("bold")


# ================== CONFUSION MATRIX (NO SAVE) ==================
def plot_confusion_matrix_preds(preds, labels, model_name="Model",
                                number_size=18, number_bold=True):
    cm = confusion_matrix(labels, preds)
    cmap, _ = SCENARIO_COLORS.get(model_name, ("Purples", None))

    fig, ax = plt.subplots(figsize=(6, 5))
    sns.heatmap(
        cm,
        annot=True, fmt='d', cmap=cmap,
        xticklabels=["Normal", "Attack"],
        yticklabels=["Normal", "Attack"],
        annot_kws={"size": number_size,
                   "weight": "bold" if number_bold else "normal"},
        cbar=False, ax=ax
    )
    ax.set_title("")
    ax.set_xlabel("")
    ax.set_ylabel("")
    style_axes(ax, tick_bold=True)
    plt.tight_layout()
    plt.show()
    plt.close(fig)


# ================== RELIABILITY DIAGRAM (NO SAVE) ==================
def plot_reliability(y_true, p_pos, bins=10, model_name="Model"):
    p_pos = np.asarray(p_pos)
    y_true = np.asarray(y_true)

    edges = np.linspace(0, 1, bins + 1)
    accs, confs = [], []

    for i in range(bins):
        left, right = edges[i], edges[i + 1]
        if i < bins - 1:
            mask = (p_pos >= left) & (p_pos < right)
        else:
            mask = (p_pos >= left) & (p_pos <= right)

        if mask.sum() == 0:
            accs.append(np.nan); confs.append(np.nan)
        else:
            accs.append(y_true[mask].mean())
            confs.append(p_pos[mask].mean())

    _, colors = SCENARIO_COLORS.get(model_name, ("Purples", ["#636EFA", "#EF553B"]))
    line_color = colors[0]

    fig, ax = plt.subplots(figsize=(6, 5))
    ax.plot([0, 1], [0, 1], '--', label='Perfect', color='gray', linewidth=2)

    valid = ~np.isnan(accs)
    ax.plot(np.array(confs)[valid], np.array(accs)[valid],
            marker='o', label=model_name, color=line_color,
            linewidth=2.5, markersize=6)

    ax.set_title("")
    ax.set_xlabel("")
    ax.set_ylabel("")
    ax.legend()
    style_axes(ax, tick_bold=True, legend_bold=True)
    plt.tight_layout()
    plt.show()
    plt.close(fig)


# ================== TRAINING CURVES (NO SAVE) ==================
def plot_training_metrics(history, model_name="Model"):
    PALETTES = {
        "CNN+BiLSTM": ["#6F42C1", "#0B3D91", "#FF7F0E", "#D62728"],
        "CNN+xLSTM":  ["#2ca02c", "#1f77b4", "#000000", "#FFD700"],
        "CNN+CoAttn": ["#1f77b4", "#0B3D91", "#FF7F0E", "#000000"],
        "HLCDA":      ["#1f77b4", "#ff7f0e", "#2ca02c", "#d62728"],  # blue, orange, green, red
    }
    colors = PALETTES.get(model_name, ["#6F42C1", "#0B3D91", "#000000", "#D62728"])

    metrics_group = ['val_accuracy', 'val_precision', 'val_recall', 'val_f1']
    epochs = range(1, len(history[metrics_group[0]]) + 1)

    styles  = ['--', '-', '-.', ':']
    markers = ['o', 's', '^', 'D']
    pretty  = {
        "val_accuracy": "Accuracy",
        "val_precision": "Precision",
        "val_recall": "Recall",
        "val_f1": "F1-Score",
    }

    fig, ax = plt.subplots(figsize=(8, 5))
    for i, m in enumerate(metrics_group):
        ax.plot(
            epochs, history[m],
            linestyle=styles[i % len(styles)],
            linewidth=2.5,
            marker=markers[i % len(markers)],
            markersize=6,
            label=pretty[m],
            color=colors[i % len(colors)]
        )
    ax.set_xlabel("Epoch", fontsize=14, fontweight='bold', color='black')
    ax.set_ylabel("Score", fontsize=14, fontweight='bold', color='black')
    ax.grid(True, linestyle=':', linewidth=0.8, alpha=0.75)
    ax.legend(loc='lower right', fontsize=12)
    style_axes(ax, tick_bold=True, legend_bold=True)
    plt.tight_layout()
    plt.show()
    plt.close(fig)

    fig, ax = plt.subplots(figsize=(8, 5))
    ax.plot(
        epochs, history['train_loss'],
        linestyle='-',
        linewidth=2.5,
        marker='o',
        markersize=6,
        label="Loss",
        color=colors[0]
    )
    ax.set_xlabel("Epoch", fontsize=14, fontweight='bold', color='black')
    ax.set_ylabel("Loss", fontsize=14, fontweight='bold', color='black')
    ax.grid(True, linestyle=':', linewidth=0.8, alpha=0.75)
    ax.legend(loc='upper right', fontsize=12)
    style_axes(ax, tick_bold=True, legend_bold=True)
    plt.tight_layout()
    plt.show()
    plt.close(fig)


# ================== t-SNE (NO SAVE) ==================
def tsne_from_pooled(Z, y, model_name="Model", perplexity=30):
    tsne = TSNE(n_components=2, perplexity=perplexity, init='pca',
                learning_rate='auto', random_state=42)
    Z2 = tsne.fit_transform(Z)

    _, colors = SCENARIO_COLORS.get(model_name, ("Purples", ["#756bb1", "#bcbddc"]))
    normal_color = colors[0]
    attack_color = "black"

    fig, ax = plt.subplots(figsize=(6, 5))
    ax.scatter(Z2[y == 0, 0], Z2[y == 0, 1], s=14, alpha=0.75, label="Normal", c=normal_color)
    ax.scatter(Z2[y == 1, 0], Z2[y == 1, 1], s=14, alpha=0.75, label="Attack", c=attack_color)

    ax.set_title("")
    ax.set_xlabel("")
    ax.set_ylabel("")
    ax.legend()
    style_axes(ax, tick_bold=True, legend_bold=True)
    plt.tight_layout()
    plt.show()
    plt.close(fig)


# ================== POOLED EMBEDDINGS ==================
@torch.no_grad()
def get_pooled_embeddings(model, loader, device=DEVICE):
    """
    Returns: (Z_all, y_all)
    Here we use logits as embeddings (consistent with your earlier code).
    """
    model.eval()
    Z_all, y_all = [], []
    for X, y, L in loader:
        X = X.to(device, non_blocking=(device=="cuda"))
        y = y.to(device, non_blocking=(device=="cuda"))
        L = L.to(device, non_blocking=(device=="cuda"))
        logits = model(X, L)
        Z_all.append(logits.detach().cpu().numpy())
        y_all.append(y.detach().cpu().numpy())
    Z_all = np.concatenate(Z_all, axis=0)
    y_all = np.concatenate(y_all, axis=0)
    return Z_all, y_all


# ============================================================
# (NEW) EVASION / ROBUSTNESS EVALUATION FOR HLCDA & BASELINES
# Uses your exact feature_names (63 CIC-style features)
# ============================================================
def build_feature_groups_exact(feature_names):
    fn = [str(x).strip().lower() for x in feature_names]

    def has_any(s, keys):
        return any(k in s for k in keys)

    groups = {
        "ports_proto": [],
        "timing": [],
        "rate": [],
        "size_bytes": [],
        "counts": [],
        "flags": [],
        "ratio_misc": []
    }

    for i, name in enumerate(fn):
        # ports/protocol
        if name in ["src port", "dst port", "protocol"]:
            groups["ports_proto"].append(i)
            continue

        # timing-like: duration, IAT, active/idle
        if has_any(name, ["duration", "iat", "active", "idle"]):
            groups["timing"].append(i)
            continue

        # rate-like: bytes/s, packets/s
        if has_any(name, ["bytes/s", "packets/s"]):
            groups["rate"].append(i)
            continue

        # flags
        if "flag count" in name:
            groups["flags"].append(i)
            continue

        # ratios/misc
        if has_any(name, ["down/up ratio"]):
            groups["ratio_misc"].append(i)
            continue

        # size/bytes/length/segment/window/header/packet size/subflow bytes
        if has_any(name, ["length", "packet length", "average packet size", "segment size",
                          "subflow", "win bytes", "header length", "packet size"]):
            groups["size_bytes"].append(i)
            continue

        # count-like totals (packets) or act data pkts
        if has_any(name, ["total fwd packet", "total bwd packets", "act data pkts"]):
            groups["counts"].append(i)
            continue

    return groups

def masked_stats_over_valid_steps(X, y, L, benign_label=0):
    """
    X: [N,T,F], y: [N], L: [N]
    stats computed over valid time steps only for y==benign_label
    """
    X = np.asarray(X); y = np.asarray(y); L = np.asarray(L)

    idx = np.where(y == benign_label)[0]
    Xb = X[idx]
    Lb = L[idx]

    Nb, T_, F_ = Xb.shape
    mask = np.zeros((Nb, T_), dtype=bool)
    for i in range(Nb):
        mask[i, :int(Lb[i])] = True

    Xflat = Xb[mask]                # [sum(L), F]
    mu = Xflat.mean(axis=0)
    sd = Xflat.std(axis=0) + 1e-8
    return mu.astype(np.float32), sd.astype(np.float32)

def apply_evasion(X, y, L, groups, benign_mu, benign_sd, recipe="timing_jitter", cfg=None, seed=42):
    rng = np.random.default_rng(seed)
    Xadv = X.copy().astype(np.float32)

    attack_idx = np.where(y == 1)[0]
    if len(attack_idx) == 0:
        return Xadv

    Na = len(attack_idx)
    _, T_, F_ = Xadv.shape

    # valid-step mask for attacks
    mask = np.zeros((Na, T_), dtype=bool)
    for j, ii in enumerate(attack_idx):
        mask[j, :int(L[ii])] = True

    def clamp_nonneg(arr, feat_idx):
        if len(feat_idx) > 0:
            arr[..., feat_idx] = np.maximum(arr[..., feat_idx], 0.0)

    cfg = cfg or {}
    alpha = float(cfg.get("alpha", 0.2))
    beta  = float(cfg.get("beta", 0.2))

    timing = groups["timing"]
    rate   = groups["rate"]
    sizeb  = groups["size_bytes"]
    counts = groups["counts"]
    ratio  = groups["ratio_misc"]

    Xa = Xadv[attack_idx]  # [Na,T,F]

    if recipe == "timing_jitter":
        if len(timing) > 0:
            noise = rng.normal(0, 1.0, size=Xa[..., timing].shape).astype(np.float32)
            Xa[..., timing] = Xa[..., timing] + (alpha * benign_sd[timing]) * noise
        clamp_nonneg(Xa, timing)

    elif recipe == "rate_shaping":
        if len(rate) > 0:
            Xa[..., rate] = Xa[..., rate] * (1.0 - alpha)
            clamp_nonneg(Xa, rate)
        if len(timing) > 0:
            noise = rng.normal(0, 1.0, size=Xa[..., timing].shape).astype(np.float32)
            Xa[..., timing] = Xa[..., timing] + (0.10 * alpha * benign_sd[timing]) * noise
            clamp_nonneg(Xa, timing)

    elif recipe == "size_frag_pad":
        if len(sizeb) > 0:
            Xa[..., sizeb] = Xa[..., sizeb] * (1.0 - alpha)
            noise = rng.normal(0, 1.0, size=Xa[..., sizeb].shape).astype(np.float32)
            Xa[..., sizeb] = Xa[..., sizeb] + (0.05 * alpha * benign_sd[sizeb]) * noise
            clamp_nonneg(Xa, sizeb)
        if len(counts) > 0:
            Xa[..., counts] = Xa[..., counts] * (1.0 + 0.15 * alpha)
            clamp_nonneg(Xa, counts)

    elif recipe == "benign_drift":
        for grp in [timing, rate, sizeb, ratio]:
            if len(grp) > 0:
                Xa[..., grp] = (1.0 - beta) * Xa[..., grp] + beta * benign_mu[grp]
        clamp_nonneg(Xa, timing + rate + sizeb + counts)

    elif recipe == "mixed":
        if len(rate) > 0:
            Xa[..., rate] *= (1.0 - alpha)
        if len(sizeb) > 0:
            Xa[..., sizeb] *= (1.0 - alpha)
        if len(timing) > 0:
            noise = rng.normal(0, 1.0, size=Xa[..., timing].shape).astype(np.float32)
            Xa[..., timing] = Xa[..., timing] + (0.10 * alpha * benign_sd[timing]) * noise
        for grp in [timing, rate, sizeb, ratio]:
            if len(grp) > 0:
                Xa[..., grp] = (1.0 - beta) * Xa[..., grp] + beta * benign_mu[grp]
        clamp_nonneg(Xa, timing + rate + sizeb + counts)

    else:
        raise ValueError(f"Unknown evasion recipe: {recipe}")

    # apply only on valid steps; keep padding unchanged
    Xa_out = Xadv[attack_idx]
    for j in range(Na):
        valid = mask[j]
        Xa_out[j, valid, :] = Xa[j, valid, :]
    Xadv[attack_idx] = Xa_out

    return Xadv

@torch.no_grad()
def eval_on_np(model, X, y, L, batch_size=SEQ_BATCH, device=DEVICE):
    ds = SeqDS(X, y, L)
    loader = DataLoader(ds, batch_size=batch_size, shuffle=False, num_workers=0, pin_memory=(device=="cuda"))

    model.eval()
    all_probs, all_preds, all_labels = [], [], []

    for Xb, yb, Lb in loader:
        Xb = Xb.to(device, non_blocking=(device=="cuda"))
        yb = yb.to(device, non_blocking=(device=="cuda"))
        Lb = Lb.to(device, non_blocking=(device=="cuda"))

        logits = model(Xb, Lb)
        probs = torch.softmax(logits, dim=1)[:, 1]
        preds = torch.argmax(logits, dim=1)

        all_probs.extend(probs.detach().cpu().numpy())
        all_preds.extend(preds.detach().cpu().numpy())
        all_labels.extend(yb.detach().cpu().numpy())

    all_probs  = np.asarray(all_probs)
    all_preds  = np.asarray(all_preds)
    all_labels = np.asarray(all_labels)

    acc  = accuracy_score(all_labels, all_preds)
    prec = precision_score(all_labels, all_preds, average="macro", zero_division=0)
    rec  = recall_score(all_labels, all_preds, average="macro", zero_division=0)
    f1   = f1_score(all_labels, all_preds, average="macro", zero_division=0)

    try:
        auroc = roc_auc_score(all_labels, all_probs)
    except Exception:
        auroc = np.nan
    try:
        auprc = average_precision_score(all_labels, all_probs)
    except Exception:
        auprc = np.nan

    return {"acc": acc, "prec": prec, "rec": rec, "f1": f1, "auroc": auroc, "auprc": auprc}

def run_evasion_suite(model, model_name, Xte, yte, Lte, Xtr, ytr, Ltr, feature_names):
    groups = build_feature_groups_exact(feature_names)
    benign_mu, benign_sd = masked_stats_over_valid_steps(Xtr, ytr, Ltr, benign_label=0)

    clean = eval_on_np(model, Xte, yte, Lte)
    print(f"\n=== EVASION SUITE: {model_name} ===")
    print(f"[CLEAN]  F1={clean['f1']:.4f}  AUPRC={clean['auprc']:.4f}  AUROC={clean['auroc']:.4f}")

    for recipe, cfg in EVASION_CFG.items():
        Xadv = apply_evasion(Xte, yte, Lte, groups, benign_mu, benign_sd,
                            recipe=recipe, cfg=cfg, seed=SEED)
        adv = eval_on_np(model, Xadv, yte, Lte)

        df1 = adv["f1"] - clean["f1"]
        dap = adv["auprc"] - clean["auprc"]

        print(f"[{recipe.upper():>12}] F1={adv['f1']:.4f} (Δ{df1:+.4f})  "
              f"AUPRC={adv['auprc']:.4f} (Δ{dap:+.4f})  AUROC={adv['auroc']:.4f}")

    return groups


# -------------------- TRAIN / EVAL (UPDATED + COSINE SCHEDULER) --------------------
def train_sequence_model(model, train_loader, val_loader, test_loader,
                         epochs=SEQ_EPOCHS, lr=LR, opt_name=OPT_NAME):
    model.to(DEVICE)
    criterion = nn.CrossEntropyLoss(weight=class_weights)
    opt = build_optimizer(opt_name, model.parameters(), lr=lr, weight_decay=WEIGHT_DECAY)

    # (A) Cosine scheduler (epoch-level)
    scheduler = None
    if SCHED_NAME.lower() == "cosine":
        scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(opt, T_max=epochs, eta_min=ETA_MIN)
    elif SCHED_NAME.lower() == "none":
        scheduler = None
    else:
        raise ValueError(f"Unknown SCHED_NAME: {SCHED_NAME}")

    history = {
        'train_loss': [],
        'train_acc': [],
        'val_accuracy': [], 'val_precision': [], 'val_recall': [], 'val_f1': [],
        'val_auroc': [], 'val_auprc': [],
        'lr_used': [],
        'lr_next': []
    }

    def eval_loader(loader, measure_latency=False, warmup_batches=5):
        model.eval()
        all_preds, all_labels, all_probs = [], [], []
        batch_times_ms = []

        with torch.no_grad():
            for b, (X, y, L) in enumerate(loader):
                X = X.to(DEVICE, non_blocking=PIN)
                y = y.to(DEVICE, non_blocking=PIN)
                L = L.to(DEVICE, non_blocking=PIN)

                if measure_latency:
                    if DEVICE.startswith("cuda"):
                        starter = torch.cuda.Event(enable_timing=True)
                        ender   = torch.cuda.Event(enable_timing=True)
                        starter.record()
                        logits = model(X, L)
                        ender.record()
                        torch.cuda.synchronize()
                        dt_ms = starter.elapsed_time(ender)
                    else:
                        t0 = time.perf_counter()
                        logits = model(X, L)
                        dt_ms = (time.perf_counter() - t0) * 1000.0

                    if b >= warmup_batches:
                        batch_times_ms.append(dt_ms)
                else:
                    logits = model(X, L)

                probs = torch.softmax(logits, dim=1)[:, 1]
                preds = torch.argmax(logits, dim=1)

                all_probs.extend(probs.detach().cpu().numpy())
                all_preds.extend(preds.detach().cpu().numpy())
                all_labels.extend(y.detach().cpu().numpy())

        all_probs  = np.asarray(all_probs)
        all_preds  = np.asarray(all_preds)
        all_labels = np.asarray(all_labels)

        acc  = accuracy_score(all_labels, all_preds)
        prec = precision_score(all_labels, all_preds, average='macro', zero_division=0)
        rec  = recall_score(all_labels, all_preds, average='macro', zero_division=0)
        f1   = f1_score(all_labels, all_preds, average='macro', zero_division=0)

        try:
            auroc = roc_auc_score(all_labels, all_probs)
        except Exception:
            auroc = np.nan
        try:
            auprc = average_precision_score(all_labels, all_probs)
        except Exception:
            auprc = np.nan

        latency = None
        if measure_latency and len(batch_times_ms) > 0:
            bt = np.asarray(batch_times_ms, dtype=np.float64)
            latency = {
                "batch_ms_mean": float(bt.mean()),
                "batch_ms_p50":  float(np.percentile(bt, 50)),
                "batch_ms_p95":  float(np.percentile(bt, 95)),
                "batch_ms_p99":  float(np.percentile(bt, 99)),
                "per_sample_ms_mean": float(bt.mean() / SEQ_BATCH)
            }

        return acc, prec, rec, f1, auroc, auprc, all_preds, all_labels, all_probs, latency

    best_f1, best_state = -1.0, None

    for ep in range(1, epochs + 1):
        model.train()
        run_loss, tot, correct = 0.0, 0, 0

        lr_used = float(opt.param_groups[0]["lr"])

        for X, y, L in train_loader:
            X = X.to(DEVICE, non_blocking=PIN)
            y = y.to(DEVICE, non_blocking=PIN)
            L = L.to(DEVICE, non_blocking=PIN)

            logits = model(X, L)
            loss = criterion(logits, y)

            opt.zero_grad()
            loss.backward()
            torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
            opt.step()

            run_loss += loss.item()
            correct += (torch.argmax(logits, 1) == y).sum().item()
            tot += y.size(0)

        v_acc, v_prec, v_rec, v_f1, v_auroc, v_auprc, _, _, _, _ = eval_loader(val_loader, measure_latency=False)

        # ---- step cosine scheduler once per epoch ----
        if scheduler is not None and SCHED_NAME.lower() == "cosine":
            scheduler.step()

        lr_next = float(opt.param_groups[0]["lr"])

        history['train_loss'].append(run_loss / len(train_loader))
        history['train_acc'].append(correct / tot if tot > 0 else 0.0)

        history['val_accuracy'].append(v_acc)
        history['val_precision'].append(v_prec)
        history['val_recall'].append(v_rec)
        history['val_f1'].append(v_f1)
        history['val_auroc'].append(v_auroc)
        history['val_auprc'].append(v_auprc)

        history['lr_used'].append(lr_used)
        history['lr_next'].append(lr_next)

        print(
            f"[SEQ] Ep {ep:02d}/{epochs} | Loss {history['train_loss'][-1]:.4f} | "
            f"TrainAcc {history['train_acc'][-1]:.4f} | ValAcc {v_acc:.4f} | "
            f"P {v_prec:.4f} | R {v_rec:.4f} | F1 {v_f1:.4f} | "
            f"AUROC {v_auroc:.4f} | AUPRC {v_auprc:.4f} | "
            f"LR_used {lr_used:.2e} | LR_next {lr_next:.2e} | OPT {opt_name} | SCHED {SCHED_NAME}"
        )

        if v_f1 > best_f1:
            best_f1 = v_f1
            best_state = {k: v.cpu() for k, v in model.state_dict().items()}

    if best_state is not None:
        model.load_state_dict(best_state)

    t_acc, t_prec, t_rec, t_f1, t_auroc, t_auprc, preds, labels, probs, latency = eval_loader(
        test_loader, measure_latency=True, warmup_batches=5
    )

    print(
        f"\n✅ SEQ Test: Acc {t_acc:.4f} | P {t_prec:.4f} | R {t_rec:.4f} | F1 {t_f1:.4f} | "
        f"AUROC {t_auroc:.4f} | AUPRC {t_auprc:.4f}"
    )
    if latency is not None:
        print(
            f"⏱ Latency (ms): batch_mean={latency['batch_ms_mean']:.3f} | "
            f"p95={latency['batch_ms_p95']:.3f} | p99={latency['batch_ms_p99']:.3f} | "
            f"~per_sample={latency['per_sample_ms_mean']:.5f}"
        )

    return model, history, preds, labels, probs


# -------------------- RUN ALL FOUR --------------------
seq_models = [
    ("CNN+BiLSTM", CNN_BiLSTM(F)),
    ("CNN+xLSTM",  CNN_xLSTM(F)),
    ("CNN+CoAttn", CNN_CoAttn(F)),
    ("HLCDA",      HLCDA(F))
]

for name, model in seq_models:
    print(f"\n🚀 Training {name} ...")
    m, hist, preds, labels, probs = train_sequence_model(
        model, tr_loader, va_loader, te_loader, epochs=SEQ_EPOCHS, lr=LR, opt_name=OPT_NAME
    )

    # 1) Confusion Matrix
    plot_confusion_matrix_preds(preds, labels, model_name=name)

    # 2) Reliability Diagram
    plot_reliability(labels, probs, bins=10, model_name=name)

    # 3) Training Curves
    plot_training_metrics(hist, model_name=name)

    # 4) t-SNE on pooled embeddings (Test set)
    Z, y_emb = get_pooled_embeddings(m, te_loader, device=DEVICE)
    tsne_from_pooled(Z, y_emb, model_name=name)

    # 5) (NEW) Evasion / Robustness evaluation
    if RUN_EVASION_EVAL:
        groups = run_evasion_suite(
            m, name,
            Xte, yte, Lte,
            Xtr, ytr, Ltr,
            feature_names
        )
        print("Feature groups used:", {k: len(v) for k, v in groups.items()})
```
